```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Step 1: Separate features and target
X = df.drop(columns=['G3'])
y = df['G3']

# Step 2: Identify categorical and numerical columns
categorical_cols = X.select_dtypes(
    include=['str', 'object', 'category']
).columns.tolist()

numerical_cols = X.select_dtypes(
    include=['number']
).columns.tolist()

# Step 3: Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])

# Step 4: Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_cols
        ),
        (
            'numerical',
            'passthrough',
            numerical_cols
        )
    ]
)

# Step 5: Create the model pipeline
model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        (
            'regressor',
            ExtraTreesRegressor(
                n_estimators=300,
                random_state=42
            )
        )
    ]
)

# Step 6: Train the model
model.fit(X_train, y_train)

# Step 7: Make predictions
y_pred = model.predict(X_test)

# Step 8: Evaluate the model
mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred) ** 0.5
r2 = r2_score(y_test, y_pred)

print("\n--- Extra Trees Model Results ---")
print(f"MAE: {mae:.3f}")
print(f"RMSE: {rmse:.3f}")
print(f"R2 Score: {r2:.4f}")
print(f"R2 Percentage: {r2 * 100:.2f}%")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import pandas as pd

models = {
    "Linear Regression": LinearRegression(),

    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        random_state=42
    ),

    "Extra Trees": ExtraTreesRegressor(
        n_estimators=300,
        random_state=42
    )
}

results = []

for name, regressor in models.items():

    # Reuse the existing preprocessing pipeline
    from sklearn.base import clone

    pipeline = clone(model)
    pipeline.set_params(regressor=regressor)

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    rmse = mean_squared_error(y_test, predictions) ** 0.5
    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": round(mae, 3),
        "RMSE": round(rmse, 3),
        "R2 Score": round(r2, 4),
        "R2 Percentage": round(r2 * 100, 2)
    })

results_df = pd.DataFrame(results)
results_df = results_df.sort_values("R2 Score", ascending=False)

results_df
```

```python
from sklearn.base import clone
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
import pandas as pd

# Remove previous-period marks
X_without_marks = df.drop(columns=['G1', 'G2', 'G3'])
y = df['G3']

# Same train-test split for fair comparison
X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X_without_marks,
    y,
    test_size=0.20,
    random_state=42
)

# Identify columns
cat_cols2 = X_train2.select_dtypes(
    include=['str', 'object', 'category']
).columns.tolist()

num_cols2 = X_train2.select_dtypes(
    include=['number']
).columns.tolist()

# Preprocessing
preprocessor2 = ColumnTransformer(
    transformers=[
        ('categorical',
         OneHotEncoder(handle_unknown='ignore'),
         cat_cols2),
        ('numerical', 'passthrough', num_cols2)
    ]
)

# Compare algorithms
models2 = {
    'Linear Regression': LinearRegression(),
    'Random Forest': RandomForestRegressor(
        n_estimators=300, random_state=42
    ),
    'Extra Trees': ExtraTreesRegressor(
        n_estimators=300, random_state=42
    )
}

results2 = []

for name, regressor in models2.items():
    pipeline = Pipeline([
        ('preprocessor', clone(preprocessor2)),
        ('regressor', regressor)
    ])

    pipeline.fit(X_train2, y_train2)
    predictions = pipeline.predict(X_test2)

    results2.append({
        'Model': name,
        'MAE': mean_absolute_error(y_test2, predictions),
        'RMSE': mean_squared_error(y_test2, predictions) ** 0.5,
        'R2 Score': r2_score(y_test2, predictions)
    })

results_without_marks = pd.DataFrame(results2)
print(results_without_marks.round(4))
```

```python
from sklearn.inspection import permutation_importance
from sklearn.base import clone
import pandas as pd

# Build model without G1 and G2
best_model = Pipeline([
    ('preprocessor', clone(preprocessor2)),
    ('regressor', ExtraTreesRegressor(
        n_estimators=300,
        random_state=42
    ))
])

# Train
best_model.fit(X_train2, y_train2)

# Permutation importance on test set
importance = permutation_importance(
    best_model,
    X_test2,
    y_test2,
    n_repeats=20,
    random_state=42,
    scoring='r2'
)

importance_df = pd.DataFrame({
    'Feature': X_test2.columns,
    'Importance': importance.importances_mean,
    'Std': importance.importances_std
}).sort_values('Importance', ascending=False)

print(importance_df.head(15))
```

```python
import joblib
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

# 1. Features and target
X = df.drop(columns=["G3"])
y = df["G3"]

# 2. Identify categorical and numerical columns
categorical_cols = X.select_dtypes(
    include=["object", "str", "category"]
).columns.tolist()

numerical_cols = X.select_dtypes(
    include=["number"]
).columns.tolist()

# 3. Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# 4. Preprocessing
preprocessor = ColumnTransformer([
    (
        "categorical",
        OneHotEncoder(handle_unknown="ignore"),
        categorical_cols
    ),
    (
        "numerical",
        "passthrough",
        numerical_cols
    )
])

# 5. Create the complete pipeline
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])

# 6. Train model
model.fit(X_train, y_train)

# 7. Save the trained model
joblib.dump(model, "student_performance_model.pkl")

print("Model trained and saved successfully!")
print("File: student_performance_model.pkl")
print("R² Score:", round(model.score(X_test, y_test), 4))
```
